# 1. Setup Catalog, Database and Volume.
1. Create a catalog - dev
2. Create a database - dev.spark_db
3. Create a volumne - dev.spark_db.datasets

```python
%sql
CREATE CATALOG IF NOT EXISTS dev;
CREATE DATABASE IF NOT EXISTS dev.spark_db;
CREATE VOLUME IF NOT EXISTS dev.spark_db.datasets;
```

```python
%run ../Utils/data-ingestion
```

3. Ingest Course Datasets

```python
DL = DataLoader()
DL.clean_ingest_data("Data", "spark_programming/data")
```

# 4. Create a table

```python
%sql
CREATE OR REPLACE TABLE dev.spark_db.diamonds(
    carat DOUBLE,
    clarity STRING,
    color STRING,
    cut STRING,
    depth STRING,
    price DOUBLE
)
```

# 5. Load Course Data

```python
%sql
INSERT INTO dev.spark_db.diamonds
SELECT * FROM json.`/Volumes/dev/spark_db/datasets/spark_programming/data/Diamonds.json`
```

# 6. Query The Table

```python
%sql
SELECT color, avg(price) AS avg_price
FROM dev.spark_db.diamonds
GROUP BY color
ORDER BY avg_price DESC
```
