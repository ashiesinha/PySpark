## Answer the following question using sf_fire_table

Create dataframe from the table to start answering.

```python
fire_df = spark.read.table("dev.spark_db.sf_fire_calls")
```

#### Q1. How many distinct types of calls were made to the fire department?

```python
%sql
select count(distinct CallType) as distinct_call_types_count
    from dev.spark_db.sf_fire_calls
```

```python

q1_df = (
    fire_df.selectExpr("count(distinct CallType) as distinct_call_types_count")
)
q1_df.show()
```

#### Q2. What were distinct types of calls made to the fire department.

```python
%sql
select distinct CallType as distinct_call_types
    from dev.spark_db.sf_fire_calls
        where CallType is not null
```

```python
q2_df = (
    fire_df.selectExpr("CallType as distinct_call_types").distinct()
)

q2_df.show()
```

#### Q3. Find out all response for delayed times greater than 5 mins?

```python
%sql
select CallNumber, Delay from dev.spark_db.sf_fire_calls
    where Delay>5
```

```python
q3_df = (
    fire_df.where("Delay > 5").
        select("CallNumber", "Delay")
        
)
q3_df.show()
```

#### Q4. What are the most common call types?

```python
%sql
select CallType, count(*) as count 
    from dev.spark_db.sf_fire_calls
        group by CallType
        order by count desc
```

```python
q4_df = (
    fire_df.select("CallType")
        .groupBy("CallType")
        .count()
        .orderBy("count", ascending=False)
)
q4_df.show()
```

#### Q5. What Zip codes accounted for most common calls.

```python
%sql
select Zipcode, count(CallNumber) as count from dev.spark_db.sf_fire_calls
    group by Zipcode
    order by count desc
```

```python
q5_df = (
    fire_df.select("Zipcode")
        .groupBy("Zipcode")
        .count()
        .orderBy("count", ascending=False)
)

q5_df.show()
```

#### Q6. What San Fransisco neighborhoods are in the zip codes 94102 and 94103

```python
%sql
select distinct Neighborhood from dev.spark_db.sf_fire_calls
    where Zipcode in (94102,94103) 
    and city = "SF"
        
```

```python
q6_df = (
    fire_df.where("Zipcode IN (94102, 94103) AND City = 'SF'")
    .select("Neighborhood")
    .distinct()
)

q6_df.show(truncate=False)
```

#### Q7. What is the sum of all calls, min. max and avg of the response times for calls?

```python
%sql
select count(CallNumber) as total_calls, avg(Delay) as avg_delay, min(Delay) as min_delay, max(Delay) as max_delay from dev.spark_db.sf_fire_calls
```

```python
q7_df = (
    fire_df.selectExpr("count(CallNumber) as total_calls", "avg(Delay) as avg_delay", "min(Delay) as min_delay", "max(Delay) as max_delay")
)
q7_df.show()
```
