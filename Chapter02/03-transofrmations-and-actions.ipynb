# 1. Answer the below questions using the dev.spark_db.sf_fire_calls

What are top 3 zip codes that accounted for most calls?

```python
%sql
-- Answer using SQL Query
select Zipcode, Calltype, count(*) as Count from dev.spark_db.sf_fire_calls where CallType is not null group by CallType, Zipcode order by count(Calltype) desc limit 3;
```

# 2. Answer using Spark Dataframe API



```python
'''
-- Answer using SQL Query
select Zipcode, Calltype, count(*) as Count  # Step 2 - Transformation 1
    from dev.spark_db.sf_fire_calls # step 1 - create dataframe
        where CallType is not null # Step 3 - Transformation 2
    group by CallType, Zipcode # Step 4 - Transformation 3
    order by count(Calltype) desc # Step 5 - Transformation 4
    limit 3; # Step 6 - Transformation 5
'''
```

```python
'''2.1 Read Data and create data frame'''
# Create a dataframe reading data from table

fire_df = spark.read.table('dev.spark_db.sf_fire_calls')
```

```python
'''2.2 Apply Transformation'''
# Transformation 1
df1 = fire_df.select("CallType", "Zipcode")

# Transformation 2
df2 = df1.where("CallType is not null")

# Transformation 3
df3 = df2.groupBy('CallType','Zipcode').count()

# Transformation 4
df4 = df3.orderBy("count", ascending=False)

# Transformation 5
df5 = df4.limit(3)
```

```python
'''2.3 Apply Action'''
# 1. Optimize the query
# 2. Executed the optimized plan
df5.show()
```

# 3. Spark Dataframe API Concepts

1. Sparks created optimized query plan
2. Dataframes are immutable
3. Every transformation returns a dataframe
4. Dataframes offer composable API

```python
#3.1

df5.explain(mode="extended") #Auxilary Method
```

```python
#3.4 Composable way of writing code

result = (
    fire_df.select("CallType", "Zipcode")
        .where("CallType is not null")
        .groupBy('CallType','Zipcode')
        .count()
        .orderBy("count", ascending=False)
        .limit(3)
)

result.show()
```
