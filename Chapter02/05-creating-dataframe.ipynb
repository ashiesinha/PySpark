# 1. Creating a dataframe using connector

```python
#reading from files.
file_df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(path="/Volumes/dev/spark_db/datasets/spark_programming/data/sf-fire-calls.csv")
)
file_df.show()
```

```python
#reading from Json.
json_df = (
    spark.read.format("json")
    .load(path="/Volumes/dev/spark_db/datasets/spark_programming/data/Diamonds.json/")
)
json_df.show()

''' CSV , Json are the data connectors for dataframe. There are other data connectors as well. Check Apache Spark Documentation'''
```

# 4. Creating a dataframe from Python List.

```python
from datetime import datetime, date

data_schema = 'id long, name string, joining_Data date, salary double'

data_list = [(1, "Ashish Sinha", date(2022, 3, 1), 100),
             (2, "Ayanshi Sinha", date(2023, 3, 1), 110)]

spark_list = spark.createDataFrame(data_list, data_schema)
spark_list.show()
```
