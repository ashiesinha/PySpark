1. Create a dataframe as below

- +---+-----------+----+---------+
- | id|       name| age|   salary|
- |100|Prashant   | 45 |    45000|
- 101   Tarun       36      33000
- 102   David       48      28000

```python
data_list = [
    (100, "Prashant", 45, 45000),
    (101, "Tarun", 36, 33000),
    (102, "David", 48, 28000)
]

data_schema = "id int, name string, age short, salary string"

# data_df = spark.createDataFrame(data_list)
# data_df = spark.createDataFrame(data=data_list).toDF("id", "name", "age", "salary")
data_df = spark.createDataFrame(data_list, data_schema)

data_df.show()
```

# 2. Add following columns to your Dataframe
    1. increment 10% of the salary up to 3000 maximum increment
    2. revised_salary: salary + increment

```python
from pyspark.sql.functions import expr

data_df_1 = (
    data_df.withColumns({
        "increment" : expr("case when (salary * 10/100) > 3000 then 3000 else (salary * 10/100) end"),
        "revised_salary" : expr("salary + increment")
    })
)
data_df_1.show()
```
